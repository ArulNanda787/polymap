```python
import requests
import pandas as pd
import json

GAMMA_API = "https://gamma-api.polymarket.com"

response = requests.get(
    f"{GAMMA_API}/markets",
    params={"limit": 5, "offset": 0},
    timeout=30
)

print("Status:", response.status_code)
print("Response time:", response.elapsed.total_seconds(), "seconds")

markets = response.json()

print("Markets returned:", len(markets))
```

```python
markets_df = pd.DataFrame(markets)

markets_df.columns.tolist()
```

```python
markets_df[
    [
        "id",
        "question",
        "conditionId",
        "slug",
        "startDate",
        "endDate",
        "volume",
        "liquidity",
        "active",
        "closed",
        "clobTokenIds"
    ]
]
```

```python
DATA_API = "https://data-api.polymarket.com"

response = requests.get(
    f"{DATA_API}/trades",
    params={
        "limit": 100,
        "offset": 0,
        "takerOnly": False
    },
    timeout=30
)

print("Status:", response.status_code)
print("Response time:", response.elapsed.total_seconds(), "seconds")

trades = response.json()

print("Trades returned:", len(trades))
```

```python
import requests

response = requests.get(
    "https://data-api.polymarket.com/trades",
    params={
        "limit": 5,
        "offset": 0,
        "takerOnly": False
    },
    timeout=30
)

print("Status:", response.status_code)
print("Content-Type:", response.headers.get("content-type"))
print(response.text[:300])
```

```python
trades = response.json()
```

```python
print(json.dumps(trades[0], indent=2))
```

```python
trades_df = pd.DataFrame(trades)

trades_df.columns.tolist()
```

```python
trade = trades[0]

print("Trade conditionId:")
print(trade["conditionId"])

print("\nTrade asset:")
print(trade["asset"])

print("\nTrade outcome:")
print(trade["outcome"])

print("\nTrade outcomeIndex:")
print(trade["outcomeIndex"])
```

```python
matching_market = markets_df[
    markets_df["conditionId"] == trade["conditionId"]
]

matching_market[
    [
        "id",
        "question",
        "conditionId",
        "outcomes",
        "outcomePrices",
        "clobTokenIds"
    ]
]
```

```python
condition_id = trades[0]["conditionId"]

response = requests.get(
    f"{GAMMA_API}/markets",
    params={
        "condition_id": condition_id
    },
    timeout=30
)

print("Status:", response.status_code)
print("Results:", len(response.json()))
```
